```python
partone = pd.read_csv(r'c:\Users\lengo\Downloads\development data\gear_target_vs_actual_1996_2000 (1).csv')
partone
```

```python
#Colapse the two value columns into one value column plus a metric label
table1 = partone.melt(id_vars=['indicator', 'year'], value_vars=['gear_target', 'actual_to_1999'],
var_name='metric', value_name='value')
```

```python
#Making the table wide
table_wide = table1.pivot(index=['metric', 'indicator'], columns='year', values='value')
table_wide = table_wide.reindex(["gear_target", "actual_to_1999"], level="metric")
table_wide
```

```python
#Removing the 2000 column
table_wide= table_wide.drop(columns=2000)
table_wide
```

```python
#Deleting the extra indicators
#Indicators to filter on
ind = ['real_gdp_growth_pct', 'real_private_investment_growth_pct', 'inflation_cpi_pct','nonagric_formal_employment_growth_pct','real_bank_rate_pct','real_bank_rate_pct', 'deficit_pct_gdp' ]

idx= pd.IndexSlice
subset = table_wide.loc[idx[:, ind], :]
subset
```
